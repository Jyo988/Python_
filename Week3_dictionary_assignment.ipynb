```python
'''Write a Python program to enter a student's name and percentage marks into a dictionary and display the information.'''
student = {}

name = input("Enter Student Name: ")
percentage = float(input("Enter Percentage Marks: "))

student["Name"] = name
student["Percentage"] = percentage

print("\nStudent Information")
print(student)
```

```python
'''Write a Python program to find the number of occurrences of each letter present in a given string.'''
text = input("Enter a string: ")

frequency = {}

for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print("\nLetter Frequency")
print(frequency)
```

```python
'''Write a Python program to find the number of occurrences of each vowel present in a given string.'''
text = input("Enter a string: ").lower()

vowels = {}

for ch in text:
    if ch in "aeiou":
        if ch in vowels:
            vowels[ch] += 1
        else:
            vowels[ch] = 1

print("\nVowel Frequency")
print(vowels)
```

```python
'''Write a Python program to accept student names and marks from the keyboard, store them in a dictionary, and display a student's marks by taking the student's name as input.'''
students = {}

n = int(input("Enter Number of Students: "))

for i in range(n):
    name = input("Enter Student Name: ")
    marks = float(input("Enter Marks: "))

    students[name] = marks

search = input("\nEnter Student Name to Search: ")

if search in students:
    print(search, "has scored", students[search], "marks.")
else:
    print("Student Not Found.")
```
