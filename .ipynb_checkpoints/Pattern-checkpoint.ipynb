```python
#Alphabet_Pyramid
n=5

for i in range(1,n+1):

    for j in range(n-i):
        print(" ",end="")

    for j in range(2*i-1):
        print(chr(64+i),end="")

    print()
```

```python
#Right_angle_triangle_with_continuous_number
n = 5

num = 1

for i in range(1, n+1):

    for j in range(1, i+1):

        print(num, end=" ")

        num += 1

    print()
```

```python
#Diamond
n = 5

# Upper

for i in range(1,n+1):

    for j in range(n-i):
        print(" ",end="")

    for j in range(2*i-1):
        print("*",end="")

    print()

# Lower

for i in range(n-1,0,-1):

    for j in range(n-i):
        print(" ",end="")

    for j in range(2*i-1):
        print("*",end="")

    print()
```

```python
#Hollow_inverted_Pyramid
n = 5

for i in range(n,0,-1):

    for j in range(n-i):
        print(" ", end="")

    for j in range(1,2*i):

        if j==1 or j==2*i-1 or i==n:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```

```python
#Hollow_square
n = 5

for i in range(1, n+1):

    for j in range(1, n+1):

        if i == 1 or i == n or j == 1 or j == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
#Mixed_Pattern
n=5

for i in range(1,n+1):

    if i%2==1:

        for j in range(1,i+1):

            print(j,end=" ")

    else:

        for j in range(1,i+1):

            print(chr(64+j),end=" ")

    print()
```

```python
#Plus
n=7

for i in range(n):

    for j in range(n):

        if i==n//2 or j==n//2:

            print("*",end=" ")

        else:

            print(" ",end=" ")

    print()
```

```python
#Right_angle_triangle_with_reverse_alphabet_order
n = 5

for i in range(1, n+1):

    for j in range(n, n-i, -1):

        print(chr(64+j), end=" ")

    print()
    
```

```python
#Reverse_triangle_with_numbers
n = 5

for i in range(n, 0, -1):

    for j in range(1, i+1):

        print(j, end=" ")

    print()
```

```python
#X
n=9

for i in range(n):

    for j in range(n):

        if i==j or i+j==n-1:

            print("*",end=" ")

        else:

            print(" ",end=" ")

    print()
```

```python
#Name
n = 5

for i in range(n):

    # J
    for j in range(n):

        if (j == 4) or (i == 4 and j < 4) or (i == 3 and j == 0):
            print("J", end="")
        else:
            print(" ", end="")

    print("   ", end="")

    # Y
    for j in range(n):

        if (i == j and i <= 2) or (i + j == 4 and i <= 2) or (j == 2 and i >= 2):
            print("Y", end="")
        else:
            print(" ", end="")

    print("   ", end="")

    # O
    for j in range(n):

        if ((i == 0 or i == 4) and (j > 0 and j < 4)) or \
           ((j == 0 or j == 4) and (i > 0 and i < 4)):
            print("O", end="")
        else:
            print(" ", end="")

    print("   ", end="")

    # T
    for j in range(n):

        if i == 0 or j == 2:
            print("T", end="")
        else:
            print(" ", end="")

    print("   ", end="")

    # I
    for j in range(n):

        if i == 0 or i == 4 or j == 2:
            print("I", end="")
        else:
            print(" ", end="")

    print()
```

```python
#Christmas_tree
sections = 3
height = 5

for s in range(sections):

    for i in range(height):
        spaces = sections * height - i - height // 2 - s
        stars = 2 * (i + s + 1) - 1

        print(" " * spaces + "*" * stars)

for i in range(4):
    print(" " * (sections * height - 2) + "|||")

```

```python
#House
roof_height = 10
house_width = 21
wall_height = 6


for i in range(roof_height):
    spaces = roof_height - i - 1
    stars = 2 * i + 1

    print(" " * spaces + "*" * stars)


for i in range(wall_height):

    if i == wall_height - 1:
        print("|" + "_" * (house_width - 2) + "|")

    elif 1 <= i <= 3:
        left_space = (house_width - 9) // 2

        print(
            "|" +
            " " * left_space +
            "|" + " " * 5 + "|" +
            " " * left_space +
            "|"
        )

    else:
        print("|" + " " * (house_width - 2) + "|")
```
