```python
%%writefile hr_report_generator/app.py

from jinja2 import Environment, FileSystemLoader
from prettytable import PrettyTable

from employee_system.employee import get_all_employees


def create_report(employee):

    environment = Environment(
        loader=FileSystemLoader("templates")
    )

    template = environment.get_template("employee_report.txt")

    report = template.render(employee=employee)

    return report


def create_table(employees):

    table = PrettyTable()

    table.field_names = [
        "ID",
        "Name",
        "Department",
        "Salary"
    ]

    for employee in employees:

        table.add_row([
            employee["id"],
            employee["name"],
            employee["department"],
            employee["salary"]
        ])

    return table


def main():

    employees = get_all_employees()

    print("HR EMPLOYEE REPORT")
    print("==================")
    print()

    for employee in employees:

        report = create_report(employee)

        print(report)
        print()

    print("EMPLOYEE TABLE")
    print("==============")

    table = create_table(employees)

    print(table)


main()
```

```python
%%writefile employee_cli/app.py

from tabulate import tabulate

from rich.console import Console
from rich.table import Table

from employee_system.employee import get_all_employees


def show_tabulate(employees):

    print("EMPLOYEE TABLE - TABULATE")
    print("=========================")

    table = tabulate(
        employees,
        headers="keys",
        tablefmt="grid"
    )

    print(table)


def show_rich(employees):

    print()
    print("EMPLOYEE TABLE - RICH")
    print("=====================")

    table = Table(title="Employee Details")

    table.add_column("ID")
    table.add_column("Name")
    table.add_column("Department")
    table.add_column("Salary")

    for employee in employees:

        table.add_row(
            employee["id"],
            employee["name"],
            employee["department"],
            str(employee["salary"])
        )

    console = Console()

    console.print(table)


def main():

    employees = get_all_employees()

    show_tabulate(employees)

    show_rich(employees)


main()
```

```python
%%writefile README.md

# Python Modules and Environments Assignment

## Project 1 - HR Report Generator

This project generates employee reports using Jinja2
and displays employee data using PrettyTable.

### Libraries

- Jinja2
- PrettyTable

## Project 2 - Employee CLI

This project displays employee information using
Tabulate and Rich.

### Libraries

- Tabulate
- Rich

## Python Modules

A module is a Python file containing reusable Python code.

Examples:

- employee.py
- salary.py
- attendance.py

## Python Package

employee_system is a package containing the employee,
salary and attendance modules.

## Virtual Environment

A virtual environment creates an isolated Python
environment for a project.

Two virtual environments are used because the two
projects have different dependencies.

## Jinja2

Jinja2 is used to create dynamic reports from templates.

## PrettyTable

PrettyTable is used to create formatted tables.

## Tabulate

Tabulate is used to display data in different table formats.

## Rich

Rich is used to create styled terminal output and tables.

## requirements.txt

requirements.txt contains the packages required by a project.

## Why use separate environments?

Separate environments prevent dependency conflicts and
keep each project's packages isolated.
```
